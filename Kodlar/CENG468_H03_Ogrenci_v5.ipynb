#### 📘 Özet: metinler (string)
- Metin tek ya da çift tırnak arasına yazılır: `'abc'` ile `"abc"` aynıdır. Birden çok satır için üç tırnak: `"""..."""`.
- Metin bir **karakter dizisidir**; her karakterin bir sırası (**indeks**) vardır ve sayma **0'dan** başlar. Eksi indeks sondan sayar: `-1` son karakterdir.
- **Dilimleme** `metin[baş:son]`: `baş` dahil, `son` **dahil değil**. `metin[:5]` ilk 5 karakter, `metin[-7:]` son 7 karakter. Aynı kural listelerde de geçerli.
- `len(metin)` uzunluğu, `"x" in metin` içinde geçip geçmediğini (True/False) verir.
- Sık kullanılan metotlar:

| Metot | Ne yapar? | Örnek → sonuç |
|---|---|---|
| `.lower()` / `.upper()` | küçük / büyük harf | `"Heart".lower()` → `'heart'` |
| `.replace(a, b)` | a'yı b ile değiştirir | `"RED".replace("R", "B")` → `'BED'` |
| `.split()` | boşluklardan böler, **liste** döndürür | `"A B C".split()` → `['A', 'B', 'C']` |
| `.strip()` | baştaki ve sondaki boşlukları atar | `" ab ".strip()` → `'ab'` |
| `.startswith(x)` | x ile mi başlıyor? | `"C536379".startswith("C")` → `True` |
| `.isnumeric()` | yalnız rakamdan mı oluşuyor? | `"99".isnumeric()` → `True` |

- Bütün metotları görmek için: `dir(str)`, bir metodun açıklaması için: `help(str.split)`.
- **Metinler değiştirilemez.** Metotlar yeni bir metin *döndürür*, eskisini değiştirmez. Kalıcı olsun istiyorsak yeniden atarız: `urun_adi = urun_adi.lower()`.

#### 📘 Özet: liste
- Liste köşeli parantezle yazılır: `x = ["a", "b", "c"]`.
- **Özellikleri:** sıralıdır (indeksle erişilir), **değiştirilebilir**, farklı tipleri bir arada tutabilir, iç içe olabilir: `x = [1, "a", True, [1, 2, 3]]`, `x[3][1]` → `2`.
- İndeks ve dilimleme kuralları metinlerle aynıdır: `x[0]`, `x[-1]`, `x[1:3]`. Olmayan bir indeks `IndexError` verir.
- Eleman değiştirme: `x[0] = 99`.
- Sık kullanılan işlemler:

| İşlem | Ne yapar? |
|---|---|
| `len(x)` | eleman sayısı |
| `x.append(e)` | sona **tek** eleman ekler (e bir listeyse, liste tek eleman olarak eklenir) |
| `x.extend(y)` | y listesinin **elemanlarını** tek tek ekler |
| `x.insert(i, e)` | i. sıraya ekler |
| `x.pop(i)` | i. elemanı çıkarır ve döndürür (i yoksa sonuncuyu) |
| `x.remove(e)` | e değerini çıkarır |
| `sorted(x)` | sıralanmış yeni liste döndürür |

- **Metot parantezle çağrılır:** `x.append(e)`. Köşeli parantez (`x.append[e]`) yazmak `TypeError` verir.

- Değiştirilebilir: `d["REG"] = 11` var olan değeri günceller, olmayan bir anahtar yazarsak yeni çift ekler.
- Python 3.7'den beri ekleme sırası korunur.
- Olmayan anahtar `KeyError` verir. Güvenli okuma: `d.get("RF", 0)` (yoksa 0 döner). Var mı diye sormak: `"REG" in d`.
- Sık kullanılan işlemler: `d.keys()` anahtarlar, `d.values()` değerler, `d.items()` (anahtar, değer) çiftleri, `d.update({"RF": 30})` ekle/güncelle, `d.pop("RF")` sil, `len(d)` çift sayısı.
- Bu derste sözlük önemli: "hangi müşteri hangi ürünü kaç adet aldı" bilgisi özünde bir sözlüktür.

### 1.9 Koşul ve döngü: iptal faturaları ayıklamak

````python
faturalar = ["536365", "536366", "C536379", "536367", "C536383"]
adet_listesi = [6, 6, -1, 32, -1]

for f in faturalar:
    if f.startswith("C"):
        print(f, "→ iptal")
    else:
        print(f, "→ satış")
````

````python
# Kodunu buraya yaz
````

#### 📘 Özet: koşul ve döngü
- **Koşul:** `if` → `elif` (istersen birden çok) → `else`. Satır sonunda **iki nokta üst üste** `:`, altındaki satırlarda **girinti** (4 boşluk) zorunludur. Girinti, hangi satırın koşula bağlı olduğunu belirler.
- **`for` döngüsü** bir koleksiyonun elemanlarını sırayla gezer: `for f in faturalar:`. Belirli sayıda tekrar için `range(n)`: `for i in range(5):` → 0, 1, 2, 3, 4.
- **Biriktirici kalıbı:** toplam, sayaç gibi değişkenler döngüden **önce** başlatılır (`toplam = 0`), döngünün içinde güncellenir (`toplam += a`, yani `toplam = toplam + a`).
- Döngü içinde koşul: `for a in adetler:` + `if a > 0:` ile yalnız istediğimiz elemanlar işlenir.
- Yardımcılar: `break` döngüyü bitirir, `continue` o turu atlar; `enumerate(x)` sıra numarasını da verir, `zip(x, y)` iki listeyi yan yana gezer.
- `while koşul:` koşul doğru olduğu sürece döner; koşul hiç yanlış olmazsa döngü bitmez.

#### 📘 Özet: fonksiyon
- Fonksiyon, belirli bir işi yapan ve tekrar tekrar çağrılabilen kod parçasıdır:
```python
def satir_tutari(adet, fiyat=1.0):
    return adet * fiyat
```
- **Parametre** tanımdaki isimdir (`adet`, `fiyat`); **argüman** çağırırken verdiğimiz değerdir (`satir_tutari(6, 2.55)`). `fiyat=1.0` **ön tanımlı** değerdir: argüman verilmezse bu kullanılır.
- **`print` ekrana yazar, `return` değer döndürür.** Döndürülen değeri bir değişkene atayıp başka hesapta kullanabiliriz. `return` yoksa fonksiyon `None` döndürür; `None * 1.2` hata verir.
- İyi fonksiyon yazımı: sık yaptığın işi fonksiyona çevir (**DRY**: kendini tekrar etme); her fonksiyon **tek bir iş** yapsın.
- Fonksiyonun açıklaması (docstring) tanımın hemen altına `""" ... """` ile yazılır, `help(fonksiyon_adi)` ile okunur.
