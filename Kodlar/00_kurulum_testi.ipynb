## Küçük bir deneme: grafik ve model

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

df = load_iris(as_frame=True).frame
sns.scatterplot(data=df, x="petal length (cm)", y="petal width (cm)", hue="target", palette="deep")
plt.title("Iris veri seti")
plt.show()

X_train, X_test, y_train, y_test = train_test_split(df.drop(columns="target"), df["target"], random_state=42)
model = DecisionTreeClassifier(random_state=42).fit(X_train, y_train)
print(f"Karar ağacı doğruluk: {model.score(X_test, y_test):.2f}")
```
